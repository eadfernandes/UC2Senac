```python
import pandas as pd
import numpy as np
df_2024_1 = pd.read_csv('Preços semestrais - AUTOMOTIVOS_2024.01.csv', sep=';', decimal=',')
df_2024_2 = pd.read_csv('Preços semestrais - AUTOMOTIVOS_2024.02.csv', sep=';', decimal=',')
df_2025_1 = pd.read_csv('Preços semestrais - AUTOMOTIVOS_2025.01', sep=';', decimal=',')
df_2025_2 = pd.read_csv('Preços semestrais - AUTOMOTIVOS_2025.02', sep=';', decimal=',')

df = pd.concat([df_2024_1, df_2024_2,
df_2025_1, df_2025_2],
ignore_index=True)
print(df.shape) 
display(df.head())

# Considerar as seguintes informações: 1) Quando mencionar Gasolina C, Entenda-se que se refere a coluna GASOLINA o banco de dados concatenado; 
# 2) O índice IQR é a diferença de quartil (veja explicação do conceito e exemplo na terceira, quarta e quinta telas anexas).
import numpy as np

# Criar arrays com os dados das colunas
array_produto = np.array(df["Produto"])
array_estado = np.array(df["Estado - Sigla"])

# Filtrar Gasolina C no Estado do Rio de Janeiro
df_gasolina_rj = df.loc[
    (array_produto == "GASOLINA") &
    (array_estado == "RJ")
].copy()

# Mostrar os primeiros registros
display(df_gasolina_rj.head())


array_var = np.array(df['GASOLINA'])
q1 = np.percentile(array_var, 25)
q3 = np.percentile(array_var, 75)
iqr = q3 - q1
limite_superior = q3 + (1.5 * iqr)
limite_inferior = q1 - (1.5 * iqr)
print(f"Limite superior: {limite_superior}")
print(f"Limite inferior: {limite_inferior}")
```
